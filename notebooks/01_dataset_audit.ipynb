# Dataset Audit

## Objective
Plan an audit of an authorised local facial-image dataset before conversion or training.

## Expected Inputs
- Completed `docs/DATASET_CARD.md` and `docs/LICENSE_LOG.md`
- An authorised, local dataset path (not stored in Git)

## Safety Checks
- Confirm licence and intended research use.
- Do not display or export facial images in this notebook.
- Record subject identifiers only if authorised and needed for person-level splits.

## Planned Cells
1. Load documented metadata.
2. Inspect file and annotation schema counts.
3. Produce a local, non-identifying audit report.
