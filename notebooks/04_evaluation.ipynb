# Evaluation

## Objective
Plan transparent evaluation of acne-lesion detection predictions.

## Expected Inputs
- Approved held-out predictions and ground truth stored locally
- Completed experiment and model-card context

## Safety Checks
- Keep evaluation images and labels out of Git.
- Report metrics, uncertainty, errors, and limitations honestly.
- Do not interpret results as medical accuracy or diagnosis.

## Planned Cells
1. Validate held-out split provenance.
2. Define documented detection metrics.
3. Review error patterns and update the model card.
