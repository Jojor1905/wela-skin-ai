# Train Baseline

## Objective
Plan a reproducible pretrained-YOLO baseline only after data governance approval.

## Expected Inputs
- Approved local YOLO configuration and person-level split manifest
- Documented model-weight licence and pinned dependencies

## Safety Checks
- Do not train before dataset licence and intended use are documented.
- Use a fixed seed and record configuration versions.
- Do not make medical or clinical claims.

## Planned Cells
1. Confirm governance prerequisites.
2. Load the approved configuration.
3. Define training command and experiment-log entry.
