# Annotation Conversion

## Objective
Plan conversion of approved source annotations to one-class YOLO labels.

## Expected Inputs
- Completed dataset, licence, and label documentation
- Authorised local source annotations

## Safety Checks
- Confirm class `0` maps only to `acne_lesion`.
- Do not infer or fabricate labels.
- Keep annotations in ignored local directories.

## Planned Cells
1. Load the documented source schema.
2. Define approved coordinate conversion.
3. Validate converted labels and report errors locally.
